 ## Data Cleaning

```scala
%python
val cleanSensorDF = sensorDF2
  .filter($"timestamp".isNotNull)
  .filter($"id".isNotNull && $"id" =!= 0)
  .filter($"temperature".isNotNull)
  .filter($"humidity".isNotNull)
  .filter($"temperature".between(0, 60))
  .filter($"humidity".between(0, 100))

println(s"Original rows: ${sensorDF2.count}")
println(s"Clean rows: ${cleanSensorDF.count}")
println(s"Removed rows: ${sensorDF2.count - cleanSensorDF.count}")
```

```scala
%python
import org.apache.spark.sql.expressions.Window
import org.apache.spark.sql.functions._

val w = Window.orderBy($"timestamp", $"id")

val cleanIntervals =
cleanSensorDF
  .filter($"timestamp".isNotNull)
  .dropDuplicates("timestamp", "temperature", "humidity")
  .withColumn("previous", lag($"timestamp", 1).over(w))
  .withColumn(
    "seconds",
    unix_timestamp($"timestamp") - unix_timestamp($"previous")
  )

cleanIntervals
  .groupBy("seconds")
  .count()
  .orderBy(desc("count"))
  .show(30, false)
```

```scala
%python
val before = cleanSensorDF.count()

val after = cleanSensorDF
  .dropDuplicates("timestamp", "temperature", "humidity")
  .count()

println(s"Exact duplicate records removed: ${before - after}")
```

```scala
%python
import org.apache.spark.sql.expressions.Window
import org.apache.spark.sql.functions._

val w = Window.orderBy("event_time")

val firstOut =
logsParsed
.withColumn(
    "gap",
    unix_timestamp($"event_time") -
    unix_timestamp(lag($"event_time",1).over(w))
)
.withColumn(
    "new_group",
    when($"gap".isNull || $"gap" > 300,1).otherwise(0)
)
.withColumn(
    "group_id",
    sum("new_group").over(w)
)

firstOut
.groupBy("group_id")
.agg(
    min("event_time").alias("first_alarm_time"),
    first("message_clean").alias("root_alarm")
)
.show(false)
```
